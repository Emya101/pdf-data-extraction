# PDF Data Extraction with PyMuPDF

**Goal:** Extract text and its position on a PDF page and organize the results into structured Python dictionaries.

This short practice notebook uses `sample-sdf-document.pdf`, a **synthetic replacement** for the original sample PDF. Its text and counts differ from the original notebook's results.

## 1. Open and inspect the document

```python
from pathlib import Path
import pymupdf

pdf_path = Path("sample-sdf-document.pdf")
if not pdf_path.is_file():
    raise FileNotFoundError(f"Sample PDF missing: {pdf_path}")

with pymupdf.open(pdf_path) as doc:
    print("Pages:", len(doc))
    print("Title:", doc.metadata.get("title", ""))
```

```python
records = []
with pymupdf.open(pdf_path) as doc:
    for page_number, page in enumerate(doc, start=1):
        for x0, y0, x1, y1, text, block, line, word_no in page.get_text("words"):
            records.append({
                "page": page_number,
                "text": text,
                "bbox": [round(x0, 2), round(y0, 2), round(x1, 2), round(y1, 2)],
                "block": block,
                "line": line,
                "word_number": word_no,
            })
print("Extracted word records:", len(records))
for record in records[:5]:
    print(record)
```

## Result and limitation

Each word now has a page reference and coordinates for layout-aware extraction. This notebook **does not perform OCR** on scanned/image-only PDFs; it expects selectable text. Use a licensed/permissioned original PDF if you want to repeat the earlier experiment.
